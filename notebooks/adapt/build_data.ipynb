```python
import pandas as pd
from utils import *
from datetime import datetime
```

```python
data_dir=Path("../../data/adapt").resolve()
experiments = [
    "full_run",
    "full_run_seed",
    "full_run_alt",
    "adapt_tuning_adapt_config_*"
]

data = {}

for experiment in experiments:
    print(experiment)
    if "*" in experiment:
        dirs = {p.stem[-1]:p for p in data_dir.glob(experiment) if (p/"scores.csv").exists()}
    else:
        ref_time = datetime.now()
        time = min([p.stem.split("_adapt")[0] for p in (data_dir/experiment).glob("*_adapt*")], key=lambda x: (ref_time-datetime.strptime(x, "%Y-%m-%d_%H:%M:%S")).total_seconds())
        dirs = {p.stem.split("_c")[1]:p for p in (data_dir/experiment).glob(f"*{time}*") if not "prepared" in p.stem and (p/"scores.csv").exists()}    
    print(dirs)
    data[experiment] = pd.concat({d:process_scores(pd.read_csv(p/"scores.csv", index_col=0)) for d, p in dirs.items()})

df = pd.concat(data, names=["experiment", "run", "design"])
df.to_csv("./figures/data.csv")
```
